```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from glcm_lib import manual_vertical_glcm_fn,manual_horizontal_glcm_fn,manual_diagonal_glcm_fn,plot_glcm,maxm_probability,energy,entropy,contrast,homogenity
```

```python
img1=cv2.imread('../assets/Picture1.jpg',0)
img2=cv2.imread('../assets/Picture2.jpg',0)
img3=cv2.imread('../assets/Picture3.jpg',0)

plt.figure(figsize=(16, 12))
plt.subplot(1, 3, 1)
plt.imshow(img1, cmap='gray')
plt.title('Picture 1')
plt.xlabel('X axis')
plt.ylabel('Y axis')

plt.subplot(1, 3, 2)
plt.imshow(img2, cmap='gray')
plt.title('Picture 2')
plt.xlabel('X axis')
plt.ylabel('Y axis')


plt.subplot(1, 3, 3)
plt.imshow(img3, cmap='gray')
plt.title('Picture 3')
plt.xlabel('X axis')
plt.ylabel('Y axis')

plt.tight_layout()
plt.show()
```

```python
p1_hori_glcm=manual_horizontal_glcm_fn(img1, levels=256)
p1_vert_glcm=manual_vertical_glcm_fn(img1, levels=256)
p1_diag_glcm=manual_diagonal_glcm_fn(img1, levels=256)
```

```python
print("Horizontal")
print(p1_hori_glcm)
```

```python
print("Vertical")
print(p1_vert_glcm)
```

```python
print("Diagonal")
print(p1_diag_glcm)
```
